📈 Detecting Outliers by IQR Method
</h3>

<div style="
    color:#1F4E79;
    background-color:#E3F2FD;
    padding:10px 18px;
    font-size:14px;
    line-height:1.6;
    border-radius:8px;
    border:1px solid #90CAF9;
    font-family:'Segoe UI', Tahoma, sans-serif;
">
    <p>
        The IQR method identifies outliers by measuring data spread between the 25th and 75th percentiles, flagging values that fall outside acceptable ranges.
    </p>
</div>

```python
df.select_dtypes(include='number').describe().round(2).T
```

```python
numeric_cols = df.select_dtypes(include='number').columns

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(col, ":", len(outliers), "outliers")
```
